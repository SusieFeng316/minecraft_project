# (2) Questions
The broad question I choose is question 1. The specific question is "Can `played_hours`, `gender`, `Age` predict whether a person subscribes to newsletter and how do these features differ between various player types in "players.csv" dataset?
The columns `played_hours`, `gender`, `Age`, and `experience` will be used to classify `subscribe`, so `select` is used to remove other columns. Since there is too little data for genders other than `Male` and `Female`, I will combine the rest and call it `Other`. To do so, gender should be in factor form.

# (3) Exploratory Data Analysis and Visualization

```r
players <- players |>
select(-hashedEmail,-name)|>
  mutate(gender = as.factor(gender)) |>
  mutate(gender = fct_collapse(
    gender,
    Male = c("Male"),
    Female = c("Female"),
    Other = c("Non-binary", "Other", "Two-Spirited", "Prefer not to say","Agender")))
players
```

```r
players |> summarize(mean(played_hours), 
                    mean_age = mean(Age,na.rm= TRUE)) 
```

```r
ggplot(players, aes(x = Age, fill = experience)) +
  geom_histogram(position = "identity", bins = 30)+ facet_grid(rows = vars(experience))+
xlab("Players' Age (year)")+
ylab("Count")+
ggtitle("Histogram of players' age versus experience")
```

Players with veteran, regular and pro experience has slightly smaller mean age than players with amateur and beginner experience.

```r
ggplot(players, aes(x=subscribe)) + geom_bar(aes(fill=gender), position="dodge")+
xlab("Subscribe")+
ylab("Count")+
ggtitle("Bar graph of gender versus whether they subscribe")
```

The distributions for `False`and `True` are similar, so gender might be a weak explanatory variable. 

```r
ggplot(players, aes(x = Age, y = played_hours, color = subscribe)) +
  geom_point()+
xlab("Age(year)")+
ylab("Played Time (hours)")+
ggtitle("Scatter plot of age versus played time in hours")
```

Played_hours and age are two predictors used to predict subscribe. As played hours increase, subscription increases. As age increase, people who subscribe decreases.
